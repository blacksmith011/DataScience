## Imbalanced Dataset Handling

###### 1. Upsampling
###### 2. Down Sampling

```python
import numpy as np 
import pandas as pd
```

```python
# set the random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classes
n_sample = 1000
class_0_ratio = 0.9
n_class_0 = int(n_sample * class_0_ratio)
n_class_1 = n_sample - n_class_0
```

```python
n_class_0,n_class_1
```

```python
class_0 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2' : np.random.normal(loc=0,scale=1,size=n_class_0),
    'target' : [0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature_1' : np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2' : np.random.normal(loc=0,scale=1,size=n_class_1),
    'target' : [1] * n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df['target'].value_counts()
```

## UpSampling

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
df_minority.head()
```

```python
df_majority.head()
```

```python
## up Sampling perform 
from sklearn.utils import resample
```

```python
df_minority_upsample = resample(df_minority,
                               replace=True,   ## Sample with replacement
                               n_samples=len(df_majority),  ## to match the majority class
                               random_state=42
                               )
```

```python
df_minority_upsample.shape
```

```python
df_minority_upsample['target'].value_counts()
```

```python
df_upsampled = pd.concat([df_majority,df_minority_upsample])
```

```python
df_upsampled['target'].value_counts()
```

## Down Sampling

```python
class_0 = pd.DataFrame({
    'feature_0' : np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_1' : np.random.normal(loc=0,scale=1,size=n_class_0),
    'target' : [0] * n_class_0
})
class_1 = pd.DataFrame({
    'feature_0' : np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_1' : np.random.normal(loc=0,scale=1,size=n_class_1),
    'target' : [1] * n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df['target'].value_counts()
```

```python
df_minority = df[df['target'] == 1]
df_majority = df[df['target'] == 0]
```

```python
df_majority_downsample = resample(df_majority,
                                 replace=True,  ## Sample with replacement
                                 n_samples=len(df_minority),  ## to match the minority class
                                 random_state=42)
```

```python
df_majority_downsample['target'].value_counts()
```

```python
df_downsampled = pd.concat([df_majority,df_majority_downsample])
```

```python
df_downsampled['target'].value_counts()
```
