```python
x = df[["Hours of Advertising per Week"]]
y = df[["Units Sold"]]
```

```python
X_train , X_test , Y_train ,Y_test =  train_test_split(x,y,test_size=0.3)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
X_test = [[6],[7]]
Y_pred = model.predict(X_test)
Y_pred
```
