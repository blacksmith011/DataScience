```python
import pandas as pd
import numpy as np
```

```python
# age = mean , height = median , city = mode , 
df = pd.DataFrame({"age":[27,28,29,'None',30,31,'None',32,33,40],
                  "height":[160,161,162,'None',163,164,165,'None',166,170],
                  "wt":[54,56,58,59,'None',60,61,'None',62,63],
                  "City":['kolkata','up','kolkata','bihar','np.na','np.na','up','kanpur','Mathura','np.na'],
                  "marks":[80,90,88,86,'None',88,92,93,'None',94]})
```

```python
df.replace(["None", "np.na"], np.nan, inplace=True)
```

```python
df.to_csv('Sample.csv',index=False)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df_drop = df.dropna()
df_drop
```

```python
df_drop.to_csv('SampleData.csv' ,index=False)
```

```python
df_drop.isnull().sum()
```

```python
df['age'].fillna(df['age'].mean(),inplace=True)
```

```python
df['height'].fillna(df['height'].median(),inplace=True)
```

```python
df['wt'].fillna(method='ffill', inplace=True)
```

```python
df['City'].fillna(df['City'].mode()[0],inplace=True)
```

```python
df['marks'].fillna(method='bfill', inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler,LabelEncoder,OneHotEncoder
```

```python
mm = MinMaxScaler()
std = StandardScaler()
```

```python
df[['mm_age','mm_height']] = mm.fit_transform(df[['age','height']])
```

```python
df[['std_wt']] = std.fit_transform(df[['wt']])
```

```python
df
```

```python
ohe = OneHotEncoder(sparse_output=False)
ohe_array = ohe.fit_transform(df[['City']])
ohe_columns = ohe.get_feature_names_out(['City'])
df_ohe = pd.DataFrame(ohe_array, columns=ohe_columns)

df_ohe
```
