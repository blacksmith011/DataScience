```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107],
    "Age": [25, 30, 28, np.nan, 45, 50, 29],
    "Weight": [58, 75, np.nan, 55, 85, 90, np.nan],
    "BloodPressure": [120, np.nan, 130, 110, 140, 150, np.nan],
    "Cholesterol": [200, 210, 220, 190, np.nan, 250, 230],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Male", "Female"]
})
```

```python
df.isnull().sum()
```

```python
df_drop = df.dropna()
df_drop
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
```

```python
df['BloodPressure'].fillna(df['BloodPressure'].mean(),inplace=True)
```

```python
df['BloodPressure'].fillna(df['BloodPressure'].mean(),inplace=True)
```

```python
df['Weight'].fillna(df['Weight'].median(),inplace=True)
```

```python
df['Cholesterol'].fillna(df['Cholesterol'].mean(),inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder,MinMaxScaler,StandardScaler
```

```python
le = LabelEncoder()
df['le_Gender'] = le.fit_transform(df['Gender'])
```

```python
df
```

```python
mm = MinMaxScaler()
std = StandardScaler()
```

```python
df[['mm_Age','mm_Weight']] = mm.fit_transform(df[['Age','Weight']])
```

```python
df[['std_Cholesterol']] = std.fit_transform(df[['Cholesterol']])
```
