```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Emp_ID": [201, 202, 203, 204, 205, 206, 207],
    "Age": [23, 29, np.nan, 35, 40, 45, 31],
    "Salary": [25000, 32000, 28000, np.nan, 50000, 60000, 35000],
    "Experience": [1, 3, 2, 5, np.nan, 8, 4],
    "Performance_Score": [3.5, np.nan, 4.0, 4.5, 4.2, np.nan, 3.8],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Male", "Female"]
})
```

```python
df.isnull().sum()
```

```python
df_drop = df.dropna()
```

```python
df_drop.to_csv('Employee_Performance.csv')
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
```

```python
df['Salary'].fillna(df['Salary'].mean(),inplace=True)
```

```python
df['Experience'].fillna(df['Experience'].mean(),inplace=True)
```

```python
df['Performance_Score'].fillna(df['Performance_Score'].mean(),inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder,MinMaxScaler,StandardScaler
```

```python
le = LabelEncoder()
```

```python
df['le_Gender'] = le.fit_transform(df['Gender'])
```

```python
mm = MinMaxScaler()
std = StandardScaler()
```

```python
df[['mm_age','mm_salary']] = mm.fit_transform(df[['Age','Salary']])
df[['std_per']] = std.fit_transform(df[['Performance_Score']])
```
